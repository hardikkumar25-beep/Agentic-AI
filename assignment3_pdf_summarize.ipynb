```python
output_folder="/content/data/extracted_data/images"
filepath="/content/data/Java Complete Notes Typed.pdf"
img_path="/content/data/extracted data/figure-1-1.jpg"
```

```python
from llama_index.core.indices import MultiModalVectorStoreIndex
from llama_index.core import SimpleDirectoryReader
from llama_index.core import StorageContext
from llama_index.vector_stores.lancedb import LanceDBVectorStore
```

```python
text_store=LanceDBVectorStore(uri="lancedb",table_name="text_collection")
img_store=LanceDBVectorStore(uri="lancedb",table_name="image_collection")
storage_context=StorageContext.from_defaults(vector_store=text_store,image_store=img_store)

docs=SimpleDirectoryReader(output_folder).load_data()
index=MultiModalVectorStoreIndex.from_documents(docs,storage_context=storage_context)
retriever=index.as_retriever(similarity_top_k=1,image_similarity_top_k=5)
```

```python
from llama_index.core.response.notebook_utils import display_source_node
from llama_index.core.schema import ImageNode

def retrieve(retriever,query_str):
  retrieval_results=retriever.retrieve(query_str)
  retrieved_img=[]
  retrieved_Text=[]

  for res_node in retrieval_results:
    if isinstance(res_node.node,ImageNode):
      retrieved_img.append(res_node.node.metadata["file_path"])
    else:
      display_source_node(res_node,source_length=200)
      retrieved_Text.append(res_node.text)
  return retrieved_img,retrieved_Text
```

```python
from PIL import Image
import matplotlib.pyplot as plt
import os
def plot_images(images_path):
  images_shown = 0
  plt.figure(figsize=(16, 9))
  for img_path in images_path:
        if os.path.isfile(img_path):
            image = Image.open(img_path)

            plt.subplot(2, 3, images_shown + 1)
            plt.imshow(image)
            plt.xticks([])
            plt.yticks([])

            images_shown += 1
            if images_shown >= 5:
                break
```

```python
query="can you tell me what is java? explain javascript?"

img,text=retrieve(retriever,query)
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain.prompts import ChatPromptTemplate
from langchain.schema.output_parser import StrOutputParser

# 1. Set up the Gemini model
# Ensure you have your Google API key set up in your environment or replace 'YOUR_API_KEY'
# gemini_llm = ChatGoogleGenerativeAI(model="gemini-pro", google_api_key="YOUR_API_KEY", temperature=0.5)
gemini_llm = ChatGoogleGenerativeAI(model="gemini-1.5-flash", temperature=0.5)


# 2. Define the prompt template
qa_tmpl_str = "You are a helpful assistant that generates quizzes. The user will provide a topic, and you will create a short mock exam quiz."

prompt = ChatPromptTemplate.from_messages(
    [
        ("system", qa_tmpl_str),
        ("human", "{query_str}"),
    ]
)

# 3. Create the chain
# The chain structure remains the same
chain = prompt | gemini_llm | StrOutputParser()

# 4. Invoke the chain with the query and print the response
resp = chain.invoke({"query_str": "Make me a mock exam quiz about the history of the internet"})
print(resp)
```

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

# LangChain-native wrapper for multimodal OpenAI
llm = ChatOpenAI(
    model="gpt-4o",
    api_key=OPENAI_API_TOKEN,
    max_tokens=200
)

prompt = ChatPromptTemplate.from_messages(
    [
        ("system", qa_tmpl_str),
        ("human", "{query_str}"),
    ]
)

chain = prompt | llm | StrOutputParser()

resp = chain.invoke({"query_str": "Make me a mock exam quiz"})
print(resp)
```
